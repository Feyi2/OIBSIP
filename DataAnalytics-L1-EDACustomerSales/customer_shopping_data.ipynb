```python
df.groupby('category')['quantity'].sum().sort_values(ascending=False).plot(kind='bar', color='skyblue', figsize=(10, 5))
plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue')
plt.show()
```

# Grouping by the category and Price with their total sums, then sorted by descending orders. Represented in a Bar Chart and Added the Chart Title on both the X and Y axes.

```python
sns.scatterplot(data=df, x='category', y='price', color='red', hue='category', palette='Set1', alpha=0.7)
plt.title('Price Distribution by Category')
plt.xlabel('Category')
plt.ylabel('Price')
plt.show()
```

```python
df1 = df.groupby(['gender', 'category'])['age'].mean().plot(kind='line', color=['blue', 'orange'], figsize=(15, 5))
plt.title('Age Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Age')
plt.xticks()
plt.show()
```
